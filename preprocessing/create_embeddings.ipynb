```python
import os
import json

DATASET_PATH = "dataset"
OUTPUT_FILE = "final_dataset.json"

final_data = []

def process_file(file_path):
    with open(file_path, "r", encoding="utf-8") as f:
        data = json.load(f)
        
        for item in data:
            # 🔥 Combine context + answers (meilleur pour embedding)
            text = item.get("context", "") + " " + " ".join(item.get("answers", []))
            
            final_data.append({
                "id": item.get("id"),
                "text": text.strip(),
                "metadata": {
                    "category": item.get("category"),
                    "tags": item.get("tags"),
                    "source": item.get("source")
                }
            })

# 🔁 Parcours récursif
for root, dirs, files in os.walk(DATASET_PATH):
    for file in files:
        if file.endswith(".json"):
            process_file(os.path.join(root, file))

# 💾 Sauvegarde
with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
    json.dump(final_data, f, indent=2, ensure_ascii=False)

print(f"Dataset fusionné : {len(final_data)} entrées")
```

```python
from sentence_transformers import SentenceTransformer
import json
import numpy as np

# 📥 Load dataset
with open("final_dataset.json", "r", encoding="utf-8") as f:
    data = json.load(f)

texts = [item["text"] for item in data]

# 📌 Model
model = SentenceTransformer("sentence-transformers/paraphrase-multilingual-mpnet-base-v2")

# ⚡ Generate embeddings
embeddings = model.encode(texts, batch_size=32, show_progress_bar=True)

# 💾 Save
np.save("embeddings.npy", embeddings)

print("Embeddings générés :", embeddings.shape)
```
